## 1. Station Initialization
Configure which devices are physical (`False`) or simulated (`True`).

```python
simulated_devices = {
    'smu': False,          # Connect to physical Keysight SMU
    'monochromator': True, # Simulate monochromator
    'rotator1': True,      # Simulate rotator 1
    'rotator2': True,      # Simulate rotator 2
    'camera': True,        # Simulate CCD camera
    'piezo150': True,      # Simulate piezo ANC150
    'piezo300': True       # Simulate piezo ANC300
}

# Initialize the QCoDeS Station
station = initialize_station(simulated=False, simulated_devices=simulated_devices)
```

## 2. Verify Connected Components
Print all connected sub-modules and instruments to verify the active configuration.

```python
print("=== Connected Components ===")
for name, inst in station.components.items():
    print(f"  - {name} ({inst.__class__.__name__})")
```

## 3. Execute Measurements

```python
# Run a 1D spectrum sweep check
plot_spectrum(station, target_wavelength=500.0, exposure_time=0.1, plotyn=True)

# Run a 2D double-gate voltage sweep experiment
run_double_gate_sweep(station, lambda_0=500.0, sample_name="gate_sweep_sample")
```
